```python
import random

def computer_guess(x):
    low = 1
    high = x
    feedback=''

    while feedback != 'c':
        if low!=high:
            guess = random.randint(low, high)
        else:
            guess = low

        feedback = input(f"Is guessed number {guess} if h,l or c")

        if feedback=='h':
            high=guess-1
            a = random.randint(low, high)
        elif feedback=='l':
            low=guess+1
            a = random.randint(low, high)
        elif feedback=='c':
            print(f"Number {guess} guessed correctly")
```

```python
import  random
def rpc():
    user= input("Enter r,p,s")
    computer = random.choice(['r','p','s'])

    if user==computer:
        print('Tie')
    else:
        if whowin(user,computer)=='u':
            print(f"User won, Computer - {computer}")
        else:
            print(f"Computer won - {computer}")

def whowin(user,computer):
    if user =='r' and computer =='s':
        return 'u'
    elif user =='p' and computer in ['r']:
        return 'u'
    elif user =='s' and computer in ['p']:
        return 'u'
    else:
        return 'c'
```

```python
def add(*args):
    a=0
    for item in args:
        a=item+a
    return int(a)

def subtract(*args):
    result=args[0]
    for item in args[1:]:
        result=result-item

    return result

def multiplier(*args):

    for item in args:
        a=item*a
    return int(a)

def devide(*args):
    try:
        for item in args:
            a=item/a
        return int(a)
    except:
        print('Right Number')
        return ''


def calculator(*args,c):
    if c not in ['+','-', '*','/']:
        print('Enter correct function')
    elif c=='+':
        output= (add(*args))
        printing_output_history(*args,output=output,c=c)
    elif c=='-':
        output= subtract(*args)
        print(output)
        printing_output_history(*args,output=output,c=c)

    elif c=='*':
        output = multiplier(*args)
        print(output)
        printing_output_history(*args,output=output,c=c)
    elif c=="/":
        output = devide(*args)
        printing_output_history(*args,output=output,c=c)

def printing_output_history(*args,output,c):
    fl = open("calc_history.txt", "a")
    a=[]
    a = [str(x) for x in args]
    res = f" {c} ".join(a)
    fl.write(f"{res} = {output} \n ")
    fl.close()
```

```python
def guess_pwd():
    word ='Amit'

    guessed_letter=''

    while True:
        guess = input('Guess the password')
        pwd = list(word)
        pwd_change=list(word)
        guessed_pwd= ['_']*len(pwd)
        guesslist = list(guess.strip())
        i=0

        if len(guesslist)>len(pwd):
                print('Incorrectly guessed password')
        else:
                while i< len(guesslist):
                    if guesslist[i] in pwd_change:
                        guessed_letter = guesslist[i]
                        pwd_index= pwd.index(guessed_letter)
                        guessed_pwd[pwd_index] = guessed_letter
                        pwd_change[pwd_index] = '_'
                    i=i+1
                x= ''.join(guessed_pwd)
                if  x==word.strip():
                    print(f'Guessed password is {''.join(guessed_pwd)}')
                    print(pwd_change)
                else:
                    print(f'Ohh, You have not guessed the full password. Guessed letters from password are {guessed_pwd}')

        if x==word.strip():
            break
        else:
            try_again = input('Do you want to try again, Y/N')
            if try_again=='N':
                break



```
