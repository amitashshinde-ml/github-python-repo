```python
def guess_pwd():
    word ='Amit'

    guessed_letter=''

    while True:
        guess = input('Guess the password')
        pwd = list(word)
        pwd_change=list(word)
        guessed_pwd= ['_']*len(pwd)
        guesslist = list(guess.strip())
        i=0

        if len(guesslist)>len(pwd):
                print('Incorrectly guessed password')
        else:
                while i< len(guesslist):
                    if guesslist[i] in pwd_change:
                        guessed_letter = guesslist[i]
                        pwd_index= pwd.index(guessed_letter)
                        guessed_pwd[pwd_index] = guessed_letter
                        pwd_change[pwd_index] = '_'
                    i=i+1
                x= ''.join(guessed_pwd)
                if  x==word.strip():
                    print(f'Guessed password is {''.join(guessed_pwd)}')
                    print(pwd_change)
                else:
                    print(f'Ohh, You have not guessed the full password. Guessed letters from password are {guessed_pwd}')

        if x==word.strip():
            break
        else:
            try_again = input('Do you want to try again, Y/N')
            if try_again=='N':
                break



```
